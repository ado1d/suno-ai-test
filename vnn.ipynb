```python
%%bash
cd /content
if [ ! -d HeartMuLa-Studio ]; then
  git clone https://github.com/fspecii/HeartMuLa-Studio.git
fi
cd HeartMuLa-Studio

# Backend deps (includes torch with CUDA, heartlib from git)
pip install -r backend/requirements.txt
```

```python
%%bash
cd /content/HeartMuLa-Studio/frontend
npm install
npm run build
```

```python
%%bash
mkdir -p /content/HeartMuLa-Studio/backend
cat > /content/HeartMuLa-Studio/backend/.env <<'EOF'
# Lyrics LLM = local Ollama (100% free, no API key)
LLM_PROVIDER=ollama
OLLAMA_HOST=http://127.0.0.1:11434
OLLAMA_MODEL=llama3.2

# Use bf16 + model swap (best mode for T4 16GB)
HEARTMULA_SEQUENTIAL_OFFLOAD=auto
HEARTMULA_COMPILE=false
EOF
cat /content/HeartMuLa-Studio/backend/.env
```

```python
%%bash
cd /content/HeartMuLa-Studio
nohup python -m uvicorn backend.app.main:app --host 0.0.0.0 --port 8000 > /tmp/backend.log 2>&1 &
echo "Backend PID: $!"
```
